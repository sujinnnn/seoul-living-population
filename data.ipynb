```python
local_people_dong['ADM_CD'] =local_people_dong['ADM_CD'].astype(str)
local_people_dong_M = local_people_dong.merge(dong[['행자부행정동코드', '행정동명']], left_on='ADM_CD', right_on='행자부행정동코드', how='left')
local_people_dong_M.rename(columns={'행정동명':'ADM_NM'}, inplace=True)
```

```python
local_people_dong_M.loc[local_people_dong_M['ADM_NM'] == '일원2동', 'ADM_NM'] = '개포3동'
```

```python
Seoul = Seoul[['geometry', 'ADM_NM']]
Seoul = Seoul.to_crs(epsg=4326)
```

```python
local_people_dong_w = local_people_dong_M.melt(
  id_vars = ['TIME', 'ADM_NM', 'TOTAL'],
  value_vars=[ 'M_10', 'M_15', 'M_20', 'M_25', 'M_30',
       'M_35', 'M_40', 'M_45', 'M_50', 'M_55', 'M_60', 'M_65', 'M_70', 'F_10',
       'F_15', 'F_20', 'F_25', 'F_30', 'F_35', 'F_40', 'F_45', 'F_50', 'F_55',
       'F_60', 'F_65', 'F_70'],
  var_name='key',
  value_name='population'
)
local_people_dong_w['GENDER'] = local_people_dong_w['key'].str[0]
local_people_dong_w['AGE'] = local_people_dong_w['key'].str[2:]
local_people_dong_w.drop(columns=['key'], inplace=True)
```
